```python
# Importing libraries
import yfinance as yf
import pandas as pd
import numpy as np
import plotly.io as pio
import plotly.express as px

# Set plot style
pio.templates.default = "seaborn"
```

```python
stock_data = yf.download('MSFT AAPL SONY META GME AMZN TSLA EC', period='max').dropna()
benchmark_data = data = yf.download('^GSPC', period='max').dropna()
```

```python
stock_data = stock_data['Close']
stock_data.head()
```

```python
benchmark_data = benchmark_data['Close']
benchmark_data.columns=['S&P 500']
benchmark_data.head()
```

```python
fig = px.line( stock_data,
              title='Stock data historical',
              labels={'value' : 'USD' , 'variable' : 'Ticker'})

fig.show()
```

```python
stock_data.info()
```

```python
fig = px.line( benchmark_data,
              labels={'value':'USD'})
fig.update_layout(title='S&P 500', showlegend= False)
fig.show()
```

```python
stock_returns = stock_data.pct_change()
fig = px.line( stock_returns,
              title='Stock % Change',
              labels={'variable':'Ticker'})
fig.show()
```

```python
sp_returns = benchmark_data['S&P 500'].pct_change()

fig = px.line(sp_returns, title='S&P 500 % Change')
fig.update_layout(showlegend=False)
fig.show()

benchmark_data.describe()
```

```python
excess_returns = stock_returns.sub(sp_returns, axis=0).dropna()

fig = px.line(excess_returns, labels={'variable':'Ticker'}, title='Excess Returns')
fig.show()
```

```python
avg_excess_returns = excess_returns.mean()
fig = px.bar(avg_excess_returns, color = avg_excess_returns.index)
fig.update_layout(title='Average excess returns', showlegend=False)
fig.show()
```

```python
std_excess_returns= excess_returns.std()
```

```python
fig = px.bar(std_excess_returns, color=std_excess_returns.index)
fig.update_layout(title='Standart deviation of excess returns', showlegend=False)
fig.show()
```

```python
daily_sharpe_ratio= avg_excess_returns.div(std_excess_returns)
#quiero sharpe ratio anual
annual_days=np.sqrt(252)
annual_sharpe_ratio=daily_sharpe_ratio.mul(annual_days)
fig = px.bar(annual_sharpe_ratio, color=annual_sharpe_ratio.index)
fig.update_layout(title='Annual sharpe ratio: Stocks vs S&P 500', showlegend=False)
fig.show()
```
